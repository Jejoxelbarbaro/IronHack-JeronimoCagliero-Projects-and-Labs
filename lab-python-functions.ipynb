```python
#1
inventory = {}
products = ["tshirt", "mug", "hat", "book", "keychain"]
def initialize_product(products):
    for product in products:
        inventory[product] = int(input (f"""Add the {product} ammount"""))
        print (f"""You added {inventory[product]} units of {product}""")
    return inventory
```

```python
#2
customer_orders = set()
def get_customer_orders ():
    input_prod = input("Type a product:").lower()
    
    while input_prod not in products:
        print ("¡NOT VALID!")
        input_prod = input("Type a product: ").lower()
    else:
        customer_orders.add(input_prod)
        print (f"""{input_prod} added to order""")
        add_prod = input ("Do you want to add a product? y/n:").lower()
        while add_prod == "y":
            input_prod = input("Type a product:").lower()
            while input_prod not in products:
                print ("¡NOT VALID!")
                input_prod = input("Type a product: ").lower()
            else:
                customer_orders.add(input_prod)
                print (f"""{input_prod} added to order""")
                add_prod = input ("Do you want to add a product? y/n:").lower()
        else:
            print ("Thanks")
        
    return customer_orders 
```

```python
#3
def update_inventory(customer_orders, inventory):
    for purchase in customer_orders:
        inventory[purchase] = inventory[purchase] - 1
    return inventory
```

```python
#4
def calculate_order_statistics (customer_orders, products):
    
    total_products_ordered = len(customer_orders)
    per_unq_prod = (total_products_ordered / len (products))*100
    order_statistics = [total_products_ordered, per_unq_prod]
    return order_statistics
```

```python
#5
def print_order_statistics (order_statistics):
    print("Total products ordered: ", order_statistics[0])
    print("Percentage of unique products ordered: ", order_statistics[1], "%")
```

```python
#6
def print_inventory (inventory):
    for product in inventory.items():
        print (product)
```

```python
#7
initialize_product(products)
get_customer_orders()
update_inventory(customer_orders, inventory)
order_statistics = calculate_order_statistics(customer_orders, products)
print_order_statistics(order_statistics)
print_inventory (inventory)
```
