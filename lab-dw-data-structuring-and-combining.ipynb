```python
import pandas as pd
import pandas as pd
```

```python
customer_data_1 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv")
customer_data_2 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file2.csv")
customer_data_3 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file3.csv")

customer_data_1.head()
```

```python
customer_data_1.info()
```

```python
customer_data_2.info()
```

```python
customer_data_3 = customer_data_3.rename(columns={
    "State":"ST",
    "Gender":"GENDER"
})
```

```python
customer_data_1["Customer Lifetime Value"] =customer_data_1["Customer Lifetime Value"].str.replace("%", "")
customer_data_2["Customer Lifetime Value"] =customer_data_2["Customer Lifetime Value"].str.replace("%", "")
```

```python
customer_data_1["Number of Open Complaints"] = (
    customer_data_1["Number of Open Complaints"]
    .str.split("/")
    .str[1]
)
customer_data_1["Number of Open Complaints"].unique()
```

```python
customer_data_2["Number of Open Complaints"] = (
    customer_data_2["Number of Open Complaints"]
    .str.split("/")
    .str[1]
)
customer_data_2["Number of Open Complaints"].unique()
```

```python
customer_data_2["Number of Open Complaints"] = customer_data_2["Number of Open Complaints"].astype("Int64")
customer_data_1["Number of Open Complaints"] = customer_data_1["Number of Open Complaints"].astype("Int64")
```

```python
customer_data_df = pd.concat([customer_data_1, customer_data_2, customer_data_3], axis = 0, ignore_index = True, join="outer")
customer_data_df.info()
```

```python
customer_data_df["Customer Lifetime Value"] = customer_data_df["Customer Lifetime Value"].astype("float64")
```

```python
customer_data_df.isnull().sum()
```

```python
customer_data_df = customer_data_df.dropna(subset=["Customer"])
customer_data_df.isnull().sum()
```

```python
customer_data_df["GENDER"] = customer_data_df["GENDER"].replace({
    
    "Femal": "F",
    "female": "F",
    "Male": "M"
})
customer_data_df["GENDER"].unique()
```

```python
customer_data_df["GENDER"] = customer_data_df["GENDER"].ffill()
```

```python
customer_data_df["ST"] = customer_data_df["ST"].replace({
    "AZ" : "Arizona",
    "WA" : "Washington",
    "Cali" : "California"
})
customer_data_df["ST"].unique()
```

```python
customer_data_df["Education"] = customer_data_df["Education"].replace({
    "Bachelors" : "Bachelor",
    "College" : "Bachelor",
    "Master" : "Graduate",
    "Doctor" : "Graduate"
})
customer_data_df["Education"].unique()
```

```python
customer_data_df["Vehicle Class"] = customer_data_df["Vehicle Class"].replace({
    "Luxury Car": "Luxury",
    "Luxury SUV": "Luxury",
    "Sports Car": "Luxury",
})
customer_data_df["Vehicle Class"].unique()
```

```python
customer_data_df.info()
```

```python
customer_data_df = customer_data_df.drop(customer_data_df.index[0])
```

```python
customer_data_df[customer_data_df["GENDER"].isna()]
```

```python
ltv_mean = customer_data_df["Customer Lifetime Value"].mean()
print(ltv_mean)
```

```python
customer_data_df = customer_data_df.fillna({"Customer Lifetime Value": ltv_mean})
```

1. You work at the marketing department and you want to know which sales channel brought the most sales in terms of total revenue. Using pivot, create a summary table showing the total revenue for each sales channel (branch, call center, web, and mail).
Round the total revenue to 2 decimal points.  Analyze the resulting table to draw insights.

```python
mkt_total_income_branch = (marketing_data_df.pivot_table(
    values="income",
    index="sales_channel",
    aggfunc="sum"
    )
    .sort_values(by="income", ascending=False)
    .astype(float).round(2)
)

mkt_total_income_branch
```

2. Create a pivot table that shows the average customer lifetime value per gender and education level. Analyze the resulting table to draw insights.

```python
mkt_gen_educ = marketing_data_df.pivot_table(
    values="customer_lifetime_value",
    index=["gender", "education"],
    aggfunc="mean"
).round(2)

mkt_gen_educ
```
